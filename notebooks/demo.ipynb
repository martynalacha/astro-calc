## 2. Planetary Weight Calculation
The weight is calculated using the formula: $W = m \cdot g$.
Thanks to `Pint`, we can use various units (kg, lbs, etc.), and the result will be automatically converted to Newtons.

```python
from astro_calc.physics import calculate_planetary_weight, ureg

# Define mass and local gravity
mass = 80 * ureg.kilogram
jupiter_gravity = 24.79 * ureg.meter / ureg.second**2

weight = calculate_planetary_weight(mass, jupiter_gravity)
print(f"An 80kg astronaut on Jupiter weighs: {weight:.2f}")
```

## 3. First Cosmic Velocity
This is the velocity needed to maintain a stable circular orbit around a planet.
The formula is: $v_1 = \sqrt{g \cdot R}$.

```python
from astro_calc.physics import calculate_first_cosmic_velocity, ureg

# Earth parameters
g_earth = 9.81 * ureg.meter / ureg.second**2
r_earth = 6371 * ureg.kilometer

v1 = calculate_first_cosmic_velocity(g_earth, r_earth)
print(f"First cosmic velocity for Earth: {v1.to(ureg.kilometer / ureg.second):.2f}")
```
